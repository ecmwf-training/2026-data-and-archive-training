### 3) This request runs but returns data for an unexpected shape. What's wrong?

The polygon below is meant to define a **triangle** over the Rhône valley in France, with vertices at (45, 3), (45, 8) and (43, 5). Polytope automatically closes a polygon ring by connecting the last point back to the first — you don't need to repeat the first point at the end of `shape`.


```python
r3 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "00",
    "expver": 1,
    "levtype": "sfc",
    "domain": "g",
    "param": "167",
    "step": "0",
    "feature": {
        "type": "polygon",
        "shape": [[45, 3], [45, 8], [43, 5], [44, 4]],
    },
}
```

```python
data_3 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r3, address="polytope.ecmwf.int", stream=False)
data_3.to_xarray()
```
