# Exercises: Area, Bounding Box and Polygon Requests

These exercises build on `polytope/area_boundingbox.ipynb`. The focus is on writing **correct Polytope requests** (not on plotting or earthkit-specific syntax), using operational forecast data (`class: od`, `stream: oper`/`enfo`) and AIFS Single (`class: ai`, `stream: oper`).

Useful references:
- [Bounding Box feature docs](https://polytope.readthedocs.io/en/latest/Service/Features/boundingbox/)
- [Polygon feature docs](https://polytope.readthedocs.io/en/latest/Service/Features/polygon/)

Each exercise gives you a request skeleton with `# TODO` markers, or a broken request for you to debug. A cell is provided after each request to actually submit it to the Polytope server — run it to check your request works, but the main goal is still getting the request dictionary right.

```python
import earthkit.data
```

### 1) There is something purposely wrong in this request. Fix it.

```python
r1 = {
"class": "od",
    "date": "-1",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "151",
    "step": 0,
    "grid": "1/1",
    "feature": {
        "type": "boundingbox",
        "points" : [[60, -20], [40, 5]]  #top left (lat lon) #bottom right (lat, lon)
    },
}
data_1 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r1, address="polytope.ecmwf.int", stream=False)
```
