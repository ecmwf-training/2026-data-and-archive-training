# Exercises: Circle Requests

These exercises build on `polytope/circle.ipynb`. The focus is on writing **correct Polytope requests** (not on plotting or earthkit-specific syntax), using operational forecast data (`class: od`, `stream: oper`/`enfo`) and AIFS Single (`class: ai`, `stream: oper`).

Useful reference: [Circle feature docs](https://polytope.readthedocs.io/en/latest/Service/Features/circle/)

Each exercise gives you a request skeleton with `# TODO` markers, or a broken request for you to debug. A cell is provided after each request to actually submit it to the Polytope server — run it to check your request works, but the main goal is still getting the request dictionary right.

```python
import earthkit.data
```

### 1) Build a basic circle request

Request **snowfall (144)** within a circle of radius 0.3° around the Zermatt ski resort (46.02N, 7.75E), for ensemble member 1, step 24.

```python
request_1 = {
    "class": "od",
    "stream": "enfo",
    "type": "pf",
    "number": "1",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "144",
    "step": "24",
    # TODO: add a "feature" dictionary of type "circle" centred on Zermatt with radius 0.3
}
```

```python
data_1 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_1, address="polytope.ecmwf.int", stream=False)
data_1.to_xarray()
```

### 2) This request fails validation. Inspect the `center` field and fix it.

```python
request_2 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "167",
    "step": "0",
    "feature": {
        "type": "circle",
        "center": [48.35, 11.79],
        "radius": 0.5,
    },
}
```
