# Exercises: Vertical Profile Requests

These exercises build on `polytope/vertical_profile.ipynb`. The focus is on writing **correct Polytope requests** (not on plotting or earthkit-specific syntax), using operational forecast data (`class: od`, `stream: oper`/`enfo`) and AIFS Single (`class: ai`, `stream: oper`).

Useful reference: [Vertical Profile feature docs](https://polytope.readthedocs.io/en/latest/Service/Features/vertical_profile/)

Each exercise gives you a request skeleton with `# TODO` markers, or a broken request for you to debug. A cell is provided after each request to actually submit it to the Polytope server — run it to check your request works, but the main goal is still getting the request dictionary right.

```python
import earthkit.data
```

### 1) Build a basic vertical profile request

Request a vertical profile of **specific humidity (133)** over **Madrid (40.4N, -3.7E)** from ensemble member 1, for pressure levels between **1000 and 100 hPa**.

```python
request_1 = {
    "class": "od",
    "stream": "enfo",
    "type": "pf",
    "number": "1",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "pl",
    "param": "133",
    "step": "0",
    # TODO: add a "feature" dictionary for a verticalprofile at Madrid, 1000 to 100 hPa
}
```

```python
data_1 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_1, address="polytope.ecmwf.int", stream=False)
data_1.to_xarray()
```

### 2) This request raises an error. Find out why and fix it.

```python
request_2 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "pl",
    "levelist": "0/to/1000",
    "param": "133",
    "step": "0",
    "feature": {
        "type": "verticalprofile",
        "points": [[38.9, -9.1]],
        "axes": "levelist",
        "range": {"start": 0, "end": 1000},
    },
}
```
