# Polytope Vertical Profile Example
## Using Ensemble Prediction System (ENS) Data from FDB

<div style="border:2px solid #f39c12; background-color:#fcf8e3; padding:10px; border-radius:5px;">
  <strong>⚠️ Warning:</strong> Operational FDB only retains data for 2 days in the past.
</div>


### Polytope Vertical Profile Feature [Documentation](https://polytope.readthedocs.io/en/latest/Service/Features/vertical_profile/)

```python
request = {
    "class": "od",
    "stream" : "enfo",
    "type" : "pf",
    "date" : -1,  # Note: date must be within the last two days 
    "time" : "0000",
    "levtype" : "pl",
    "expver" : "0001", 
    "domain" : "g",
    "param" : "133", # Specific humidity (q)
    "number" : "1",
    "step" : "0",
    "feature" : {
        "type" : "verticalprofile",
        "points": [[38.9, -9.1]],
        "axes": "levelist",
        "range" : {
            "start" : 0,
            "end" : 1000,
        }
    },
}

data = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request, stream=False, address='polytope.ecmwf.int')
```

```python
dx = data.to_xarray()
dx
```

```python
import earthkit.plots
chart = earthkit.plots.Subplot(figsize=(8, 8))

chart.line(dx, x="q")

chart.title("Vertical profile of {variable_name} over Lisbon from the DestinE's Climate DT")

chart.ylabel("hPa")
chart.xlabel("{units}")

chart.ax.invert_yaxis()

chart.show()
```
