### Polytope Circle Feature [Documentation](https://polytope.readthedocs.io/en/latest/Service/Features/circle/)

### Airport Weather Watch: low-level wind variability around Munich Airport

Here we extract the **10m wind speed** (derived from its u/v components) within a circle of radius 0.75° centred on Munich Airport (MUC), and show that it varies by a factor of 3 or more across the circle — information a single-point query would miss entirely.

```python
request = {
    "class": "ai",
    "model": "aifs-single",
    "date": "-3",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "165/166",  # 10m wind components (u, v)
    "step": "12",
    "feature": {
        "type": "circle",
        "center": [[48.35, 11.79]],  # Munich Airport (MUC)
        "radius": 0.75,
    },
}
data = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request, address="polytope.ecmwf.int", stream=False)
```

```python
dx = data.to_xarray()
dx["speed"] = (dx["10u"] ** 2 + dx["10v"] ** 2) ** 0.5

print(dx)
```

```python
chart = earthkit.plots.Map(domain=[9, 14, 46, 50])
chart.scatter(
    dx,
    x="longitude",
    y="latitude",
    z="speed",
    units="m/s",
    s=80,
)
chart.coastlines()
chart.borders()
chart.gridlines()
chart.legend()
chart.title("10m wind speed within 0.75\u00b0 of Munich Airport (MUC)")
chart.show()
```
