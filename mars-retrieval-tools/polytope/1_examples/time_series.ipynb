```python
import earthkit.data
from earthkit.plots.interactive import Chart
```

# Polytope Example for Time Series
## Using data from AIFS Single

<div style="border:2px solid #f39c12; background-color:#fcf8e3; padding:10px; border-radius:5px;">
  <strong>⚠️ Warning: </strong> AIFS FDB only retains data for up to 3-5 days in the past.
</div>


### Polytope Time Series Feature [Documentation](https://polytope.readthedocs.io/en/latest/Service/Features/timeseries/)

```python
request = {
    "class": "ai",
    "model": "aifs-single",
    "date": "0",  # AIFS FDB only retains data for 3-5 days in the past
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "165/167",
    "feature": {
        "type" : "timeseries",
        "points": [[50.74,7.10]],
        "axes": "step",
        "range": {"start" : 0, "end": 240}
    }
}
data = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request, address="polytope.ecmwf.int", stream=False)
```

```python
dx = data.to_xarray()
dx
```

```python
# convert temperature from Kelvin to Celsius
dx['2t'] = dx['2t'] - 273.15
dx['2t'].attrs['units'] = '°C'
dx
```

```python
TIME_FREQUENCY = "6h"

chart = Chart()
chart.title(f"ECMWF ensemble meteogram at 50.74°N, 7.10°E (Bonn)")

chart.line(
    dx,
    line_color='grey',
    time_frequency=TIME_FREQUENCY,
)

chart.show()
```
