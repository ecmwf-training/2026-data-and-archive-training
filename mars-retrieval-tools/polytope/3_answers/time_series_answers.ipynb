```python
import earthkit.data
```

### 1) Solution

```python
request_1 = {
    "class": "ai",
    "model": "aifs-single",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "151",
    "date": "-3",
    "feature": {
        "type": "timeseries",
        "points": [[64.15, -21.95]],
        "axes": "step",
        "range": {"start": 0, "end": 72},
    },
}
```

```python
data_1 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_1, address="polytope.ecmwf.int", stream=False)
data_1.to_xarray()
```

### 2) Fix

`step` and `range` both define the time axis — this is overspecified and Polytope raises an error. Only one of them should be used; since `range` is already set inside `feature`, the top-level `step` key must be removed.

```python
request_2 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "167",
    "feature": {
        "type": "timeseries",
        "points": [[41.9, 12.5]],
        "axes": "step",
        "range": {"start": 0, "end": 240},
    },
}
```

```python
data_2 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_2, address="polytope.ecmwf.int", stream=False)
data_2.to_xarray()
```

### 4) Fix

`"date": "-6"` is outside the AIFS FDB's 3-5 day retention window. Changing it to e.g. `"-3"` fixes the request.

```python
request_4 = {
    "class": "ai",
    "model": "aifs-single",
    "date": "-3",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "167",
    "feature": {
        "type": "timeseries",
        "points": [[50.74, 7.10]],
        "axes": "step",
        "range": {"start": 0, "end": 48},
    },
}
```

```python
data_4 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_4, address="polytope.ecmwf.int", stream=False)
data_4.to_xarray()
```

### 5) Solution

The level is set via `levtype`/`levelist` in the main request body, not inside `feature`.

```python
request_5 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "param": "133",
    "levtype": "pl",
    "levelist": "850",
    "feature": {
        "type": "timeseries",
        "points": [[40.4, -3.7]],
        "axes": "step",
        "range": {"start": 0, "end": 48},
    },
}
```

```python
data_5 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_5, address="polytope.ecmwf.int", stream=False)
data_5.to_xarray()
```

### 6) Fix

`points` order matters: it must be `[latitude, longitude]`. The original `[7.10, 50.74]` was swapped and would be interpreted as 7.10N, 50.74E — a point in the North Sea rather than Bonn.

```python
request_6 = {
    "class": "ai",
    "model": "aifs-single",
    "date": "-3",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "167",
    "feature": {
        "type": "timeseries",
        "points": [[50.74, 7.10]],
        "axes": "step",
        "range": {"start": 0, "end": 48},
    },
}
```

```python
data_6 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_6, address="polytope.ecmwf.int", stream=False)
data_6.to_xarray()
```
