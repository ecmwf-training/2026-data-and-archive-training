# Answers: Polygon Requests

Solutions for `exercises/polygon_exercises.ipynb`.


```python
import earthkit.data
```

### 1) Solution

`shape` takes a list of polygons when more than one is required. This request returns 2 polygons × 2 members × 2 steps = **8** individual results.

```python
r1 = {
    "class": "od",
    "stream": "enfo",
    "type": "pf",
    "date": "-1",
    "time": "00",
    "expver": 1,
    "levtype": "sfc",
    "domain": "g",
    "param": "228",
    "number": "1/2",
    "step": "0/6",
    "feature": {
        "type": "polygon",
        "shape": [
            [[43, -8], [43, 3], [36, -3], [43, -8]],
            [[46, 7], [46, 18], [38, 13], [46, 7]]
        ]
    },
}

# 2 polygons x 2 members x 2 steps = 8 combinations
```

```python
data_1 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r1, address="polytope.ecmwf.int", stream=False)
data_1.to_xarray()
```

### 2) Fix

A polygon needs **at least 3 vertices** to enclose an area — two points only define a line, which isn't a valid polygon. Adding a third point fixes the request.


```python
r2 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "00",
    "expver": 1,
    "levtype": "sfc",
    "domain": "g",
    "param": "167",
    "step": "0",
    "feature": {
        "type": "polygon",
        "shape": [[45, 5], [50, 10], [48, 2]],
    },
}
```

```python
data_2 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r2, address="polytope.ecmwf.int", stream=False)
data_2.to_xarray()
```

### 3) Solution

Polytope assumes the first point in `shape` equals the last and closes the ring automatically — you should not add an explicit closing point. The original request's 4th point `(44, 4)` does not match the first point `(45, 3)`, so instead of closing the intended triangle it adds a real 4th vertex, turning the polygon into an unintended quadrilateral. Removing that extra point fixes it.


```python
r3 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "00",
    "expver": 1,
    "levtype": "sfc",
    "domain": "g",
    "param": "167",
    "step": "0",
    "feature": {
        "type": "polygon",
        "shape": [[45, 3], [45, 8], [43, 5]],
    },
}
```

```python
data_3 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r3, address="polytope.ecmwf.int", stream=False)
data_3.to_xarray()
```

### 4) Solution

A single polygon shape with 2 params × 3 steps returns 1 × 2 × 3 = **6** individual results.


```python
r4 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "00",
    "expver": 1,
    "levtype": "sfc",
    "domain": "g",
    "param": "167/151",
    "step": "0/6/12",
    "feature": {
        "type": "polygon",
        "shape": [[52, 2], [52, 7], [49, 4]],
    },
}

# 1 polygon x 2 params x 3 steps = 6 combinations
```

```python
data_4 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r4, address="polytope.ecmwf.int", stream=False)
data_4.to_xarray()
```
