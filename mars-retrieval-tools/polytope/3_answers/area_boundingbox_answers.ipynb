### 1) Fix

Bounding box features do not support server-side regridding, so the `"grid": "1/1"` key must be removed (it is only valid for full-field requests).

```python
r1 = {
    "class": "od",
    "date": "-1",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "151",
    "step": 0,
    "feature": {
        "type": "boundingbox",
        "points": [[60, -20], [40, 5]]
    },
}
```

```python
data_1 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r1, address="polytope.ecmwf.int", stream=False)
```

### 2) Answer

The first request fails because feature requests can only retrieve data from the FDB, and the operational FDB only retains data for 2 days in the past (here the request asks for `date: -5`).
The second request works because it's a full (non-feature) request, so data can be fetched not only from the FDB but also from the MARS archive.

A corrected version of the first request could be:

```python
r2 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "00",
    "expver": 1,
    "levtype": "sfc",
    "param": "167",
    "step": 24,
    "feature": {
        "type": "boundingbox",
        "points": [[60, 4], [55, 32]]
    },
}
```

```python
data_2 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r2, address="polytope.ecmwf.int", stream=False)
```

### 3) Solution

Fill in the `feature` dictionary with a bounding box over Scandinavia, keeping the rest of the skeleton unchanged.

```python
r3 = {
    "class": "od",
    "stream": "enfo",
    "type": "pf",
    "number": "1/to/5",
    "date": "-1",
    "time": "00",
    "expver": 1,
    "levtype": "sfc",
    "domain": "g",
    "param": "167",
    "step": 24,
    "feature": {
        "type": "boundingbox",
        "points": [[60, 4], [55, 32]] # 60N, 4E, 55N, 32E
    },
}
```

```python
data_3 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", r3, address="polytope.ecmwf.int", stream=False)
```

### 4) Solution

```python
request_4 = {
    "class": "ai",
    "model": "aifs-single",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "151",
    "date": "-3",
    "feature": {
        "type": "boundingbox",
        "points": [[60, -20], [40, 5]]
    },
}
```

```python
data_4 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_4, address="polytope.ecmwf.int", stream=False)
```

Although both requests cover the same area, the full request reads the whole global field first and then interpolates/crops it to the requested area, whereas the feature request reads only the selected area directly from the FDB. This is why the feature request completes faster.
