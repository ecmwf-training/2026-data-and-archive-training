# Answers: Vertical Profile Requests

Solutions for `exercises/vertical_profile_exercises.ipynb`.

```python
import earthkit.data
```

### 1) Solution

```python
request_1 = {
    "class": "od",
    "stream": "enfo",
    "type": "pf",
    "number": "1",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "pl",
    "param": "133",
    "step": "0",
    "feature": {
        "type": "verticalprofile",
        "points": [[40.4, -3.7]],
        "axes": "levelist",
        "range": {"start": 100, "end": 1000},
    },
}
```

```python
data_1 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_1, address="polytope.ecmwf.int", stream=False)
data_1.to_xarray()
```

### 2) Fix

Both `levelist` (main body) and `range` (inside `feature`) define the vertical axis — this is overspecified. Since `range` is already set inside `feature`, the top-level `levelist` key must be removed.

```python
request_2 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "pl",
    "param": "133",
    "step": "0",
    "feature": {
        "type": "verticalprofile",
        "points": [[38.9, -9.1]],
        "axes": "levelist",
        "range": {"start": 0, "end": 1000},
    },
}
```

```python
data_2 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_2, address="polytope.ecmwf.int", stream=False)
data_2.to_xarray()
```

### 4) Solution

When `levelist` is given explicitly in the main body, `range` is omitted from `feature` (only one of the two is allowed).

```python
request_4 = {
    "class": "ai",
    "model": "aifs-single",
    "date": "-3",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "pl",
    "param": "129",
    "step": "0",
    "levelist": "1000/850/700/500/250",
    "feature": {
        "type": "verticalprofile",
        "points": [[59.9, 10.75]],
        "axes": "levelist",
    },
}
```

```python
data_4 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_4, address="polytope.ecmwf.int", stream=False)
data_4.to_xarray()
```
