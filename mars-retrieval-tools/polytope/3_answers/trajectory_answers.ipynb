### 1) Solution

```python
request_1 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "049",
    "step": "0",
    "feature": {
        "type": "trajectory",
        "points": [[51.13, 1.31], [50.85, 1.20], [50.97, 1.85]],
        "axes": ["latitude", "longitude"],
        "inflation": 0.2,
        # "inflate" : 'round', # no need to specify since default is round

    },}
```

### 2) Fix

A trajectory needs **at least two points** to define a path — a single point isn't a valid trajectory (use the `verticalprofile` or `timeseries` feature for a single location). Adding a second waypoint along the trail fixes the request.

```python
request_2 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "165/166",
    "step": "0",
    "feature": {
        "type": "trajectory",
        "points": [[46.02, 7.75], [46.00, 7.80]],
        "axes": ["latitude", "longitude"],
        "inflation": 0.1,

    },}
```

```python
data_2 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_2, address="polytope.ecmwf.int", stream=False)
data_2.to_xarray()
```

### 3) Solution

The route is narrow across the valley but extends along it, so the box inflation uses a small latitude half-width and a larger longitude half-width.

```python
request_3 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "167",
    "step": "0",
    "feature": {
        "type": "trajectory",
        "points": [[47.26, 11.40], [47.58, 12.17]],
        "axes": ["latitude", "longitude"],
        "inflate": "box",
        "inflation": [0.05, 0.3],
    },
}
```

```python
data_3 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_3, address="polytope.ecmwf.int", stream=False)
data_3.to_xarray()
```

### 4) Fix

With `"inflate": "box"` and 3 axes (`latitude`, `longitude`, `step`), `inflation` must be a list of 3 values, one per axis — a single number like `0.3` is ambiguous and rejected.

```python
request_4 = {
    "class": "ai",
    "model": "aifs-single",
    "date": "-3",
    "expver": 1,
    "stream": "oper",
    "time": "00",
    "type": "fc",
    "levtype": "sfc",
    "param": "167",
    "feature": {
        "type": "trajectory",
        "points": [[54.0, 9.0, 0], [52.0, 10.0, 24], [50.0, 11.0, 48]],
        "axes": ["latitude", "longitude", "step"],
        "inflate": "box",
        "inflation": [0.3, 0.3, 6],
    },
}
```

```python
data_4 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_4, address="polytope.ecmwf.int", stream=False)
data_4.to_xarray()
```
