# Answers: Circle Requests

Solutions for `exercises/circle_exercises.ipynb`.

```python
import earthkit.data
```

### 1) Solution

```python
request_1 = {
    "class": "od",
    "stream": "enfo",
    "type": "pf",
    "number": "1",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "144",
    "step": "24",
    "feature": {
        "type": "circle",
        "center": [[46.02, 7.75]],
        "radius": 0.3,
    },
}
```

```python
data_1 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_1, address="polytope.ecmwf.int", stream=False)
data_1.to_xarray()
```

### 2) Fix

`center` must be a **nested list** containing a single `[lat, lon]` point, i.e. `[[48.35, 11.79]]`, not a flat pair.

```python
request_2 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "167",
    "step": "0",
    "feature": {
        "type": "circle",
        "center": [[48.35, 11.79]],
        "radius": 0.5,
    },
}
```

```python
data_2 = earthkit.data.from_source(
    "polytope", "ecmwf-mars", request_2, address="polytope.ecmwf.int", stream=False)
data_2.to_xarray()
```

### 3) Fix

```python
request_3 = {
    "class": "od",
    "stream": "oper",
    "type": "fc",
    "date": "-1",
    "time": "0000",
    "expver": "0001",
    "domain": "g",
    "levtype": "sfc",
    "param": "049",
    "step": "6",
    "feature": {
        "type": "circle",
        "center": [[45.46, 9.19]],
        "radius": 0.4,
    },
}
```
